```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
```

```python
df = sns.load_dataset("titanic")
df
```

```python
df[["sex","class","embark_town"]].isna().sum() # Boş Veriler incelendi
```

yukarıdaki kod incelendiğinde embark_town da 2 tane verinin boş olduğu gözlemlendi. 
Boş veri sayısı az olduğu için silinmesi tercih edildi.
Veri silmek için iki tane kod yazılabilir bunlar;
    df["embark_town"].dropna()
    df.dropna(subset=['embark_town'])


```python
df = df.dropna(subset=["embark_town"])
df
```

```python
df[["sex","class","embark_town"]].isna().sum()
```

```python
df["sex"].value_counts()
```

```python
df["embark_town"].value_counts()
```

```python
df["class"].value_counts()
```

```python
df_onehot = pd.get_dummies(df,columns=["sex","embark_town"],drop_first=True)
```

```python
df_onehot
```

# LABEL ENCODER

```python
label_encoder = LabelEncoder()
```

```python
df_label = df.copy() 
df_label
```

```python
df_label["sex_encoded"] = label_encoder.fit_transform(df_label["sex"])
df_label
```

# ORDİNAL ENCODER

```python
df
```

```python
df_ordinal = df.copy()
```

```python
class_order = ["Third","Second","First"]
```

```python
ordinal_encoder = OrdinalEncoder(categories=[class_order])
```

```python
df_ordinal["class"] = ordinal_encoder.fit_transform(df_ordinal[["class"]])
df_ordinal
```

```python
df
```

```python
fig , axes = plt.subplots(1,3,figsize=(15,5))
df["sex"].value_counts().plot(kind="bar", ax=axes[0], title="Original Categorical")
df_label["sex"].value_counts().plot(kind="bar", ax=axes[1], title="Label Encoding")
df_onehot["sex_male"].value_counts().plot(kind="bar", ax=axes[2], title="One_hot")
```
